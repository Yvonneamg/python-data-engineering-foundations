## Patient demographics

```python
# Hospital patient records
patients = [
    {"patient_id": "P001", "age": 45, "sex": "Female", "condition": "Diabetes"},
    {"patient_id": "P002", "age": 62, "sex": "Male", "condition": "Hypertension"},
    {"patient_id": "P003", "age": 31, "sex": "Female", "condition": "Asthma"},
    {"patient_id": "P004", "age": 54, "sex": "Male", "condition": "Diabetes"}
]
patients
```

```python
# Printing all patients IDs
for patient in patients:
    print(patient["patient_id"])
```

```python
#Calculating the total number of patients
print("Total number of patients:", len(patients))
```

```python
# Calculating the average age of patients
average_age = sum(patient["age"] for patient in patients) / len(patients)
print("Average age of patients:", average_age)
```

```python
# Identifying patients with Diabetes
for patient in patients:
    if patient["condition"] == "Diabetes":
        print(f"Patient {patient['patient_id']} has Diabetes.")
```

```python
# Count how may patients are male and how many are female
male_count = sum(1 for patient in patients if patient["sex"] == "Male")
female_count = sum(1 for patient in patients if patient["sex"] == "Female")
print(f"Number of male patients: {male_count}")
print(f"Number of female patients: {female_count}")
```

```python
# Add a new patient record
new_patient = {"patient_id": "P005", "age": 29, "sex": "Female", "condition": "Hypertension"}
patients.append(new_patient)
```

```python
print("Updated patient records:")
for patient in patients:
    print(patient)
```

## Clinical Measurements

```python
glucose_readings = [85, 110, 145, 98, 180, 72, 130, 105]
```

```python
print("glucose_readings:", glucose_readings)
```

```python
# Calculating the mean glucose level
mean_glucose = sum(glucose_readings) / len(glucose_readings)
print("Mean glucose level:", mean_glucose)
```

```python
# Calculating the maximum and minimum glucose levels
max_glucose = max(glucose_readings)
min_glucose = min(glucose_readings)
print("Maximum glucose:", max_glucose)
print("Minimum glucose:", min_glucose)
```

```python
# Checking glucose readings against thresholds
for glucose in glucose_readings:
    if glucose > 140:
        print(f"Glucose reading {glucose} is high.")
    elif glucose < 70:
        print(f"Glucose reading {glucose} is low.")
    else:
        print(f"Glucose reading {glucose} is normal.")
       
```

```python
# Counting the number of glucose readings greater than or equal to 126
glucose_reading_greater_than_126 = sum(1 for glucose in glucose_readings if glucose >= 126)
print(f"Number of glucose readings greater than or equal to 126: {glucose_reading_greater_than_126}")
```

```python
# Counting the number of glucose readings less than 70
glucose_reading_less_than_70 = sum(1 for glucose in glucose_readings if glucose < 70)
print(f"Number of glucose readings less than 70: {glucose_reading_less_than_70}")
```
